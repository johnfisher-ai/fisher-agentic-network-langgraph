# Agentic Travel Partner Agent Network

## Goals
A multi-agent and MCP server network for Ofishal Partners would fundamentally transform how travel suppliers (hotels, vacation rentals, airlines) interact with the Online Travel Agency (OTA) platform. Instead of navigating static dashboards, partners would engage with a dynamic ecosystem of specialized AI agents that proactively collaborate to optimize revenue, enhance guest experiences, and streamline operations.

**Ofishal** is a fictitious travel, planning, and booking company with the tagline "Your *Ofishal* Travel Partner".

## Core Business Values
The core business value lies in:

**Proactive Optimization:** Agents don't just report data; they suggest and execute actions (e.g., adjusting rates during a sudden demand spike).

**Hyper-Personalization:** Tailoring strategies to the specific partner's size, location, and market segment.

**Operational Efficiency:** Automating repetitive tasks like review responses and content updates.

**Risk:** Score the risk of implementing the change. High risk requires a Human-in-the-Loop.

## Agent Roles

**Gatherer:** Primarily collects data from internal and external systems. E.g., a Scout Agent would look for events in your area (concerts, sports, construction); the Official Aviation Guide (OAG) would collect inbound airline traffic trends; and a Competitive Intelligence Agent would collect pricing trends, amenities, offerings, and bundles from your competitors.

**Analyst:** Determines the significance of data collected using sound statistical data processing.

**Strategist:** Synthesizes the Gatherer and Analyst inputs and provides a comprehensive plan to move Key Performance Indicators (KPIs) for your business. E.g., Average Daily Rate (to increase this you need to do x, y, z); Booking trends (to combat downtrends you need to bundle a family plan and update your content to include the following tags and filters to ensure high Search Engine Optimization).

**Actionable:** If low risk, have the AI call appropriate MCPs and APIs to implement the Strategy (e.g., BookingRatesMCP, ContentManagerMCP, seoMCP, etc.).

**Risk:** Determines the risk of the change on a scale of 1 to 10; anything above 7 requires a human-in-the-loop to approve the change via MCP and APIs, or a human manually implements the strategy.

**HumanInTheLoop:** Interactive human data gathering, e.g., human response to YES/NO questions, value questions, etc.

## Agents (In this prototype)
**Revenue Strategist Agent**: Maximize revenue per available room (RevPAR) and occupancy by analyzing real-time market data and competitor rates.

**Content and Visibility Agent**: Ensure the property listing is optimized for Ofishal sort algorithem and SEO, improving conversion rates.

**Guest Experience and Reputation Agent**: Manages brand reputation by analyzing reviews, automating responses, and flagging service issues before they affect the partner's score.

**Market Intelligence Scout Agent**:  Monitors macro-travel trends, flight capacity dta, and local events to predict demand shifts before they happen.

**Human Approval Agent**:  Manages human-in-the-loop interactions and responds accordinaly.

**Broker**: Acts as the intelligent interafce between the human partner and the specialized agents.  The user does not need to *talk* to teh agents individually; the user states high-level busienss problem, and the Broker decomposes the task and delegates it.

## Broker

In an agentic network, the Broker (or "Orchestrator") acts as the intelligent interface between the human partner and the specialized agents. The user does not need to talk to the agents individually; they state a high-level business problem, and the Broker decomposes the task and delegates it.

**Role**: Project Manager and Strategist. It understands the user's intent, breaks it down into sub-tasks, assigns them to the correct agents, synthesizes the results, and presents a cohesive business solution.

**Instructions**:

You are the Ofishal Partner Strategy Lead. Your goal is to solve complex business problems for partners.
- *Analyze* the user's request.
- *Delegate* sub-tasks to the Revenue Strategist, Content & Visibility, Guest Experience and Reputation, or Market Intelligence Scout agents.
- *Wait* for their outputs.
- *Synthesize* their findings into a single, actionable strategic plan.
- *Ask* for permission to execute actions via the specialized agents.

### Collaboration Scenario: Low Season Slump

**User Prompt to Broker**:

"I'm worried about November. Bookings are pacing 20% behind last year. What is going on and how do we fix it?"

**Step 1: Broker Decomposition & Delegation** The Broker analyzes the request and triggers three agents in parallel (hidden from user view):

- To Market Scout: "Check specifically for demand drops or event cancellations in our region for November."

- To Revenue Strategist: "Analyze our November rates vs. comp set. Are we overpriced?"

- To Content Agent: "Check our listing visual appeal for 'Autumn/Winter' relevance."

**Step 2: Agent Responses (Internal Network)**

- Market Scout: "Major tech conference usually held in November moved to October. Demand is down, but leisure travel is steady."

- Revenue Strategist: "We are priced at $250 (conference rates), but competitors have already dropped to $190 (leisure rates)."

- Content Agent: "Our main photo still shows the pool, which is closed in November. We lack 'cozy' keywords."

**Step 3: Broker Synthesis & Response** The Broker combines these isolated insights into a strategy.

**Final Response to Partner**:

"I have analyzed the situation and found the root cause. The Tech Summit moved to October, so we are missing that corporate demand. However, we can pivot to capture leisure travelers.

**The Strategy**:

- Pricing (Revenue Agent): You are currently priced for business travelers ($250). I recommend dropping rates to $195 to undercut the leisure market comp set.

- Merchandising (Content Agent): Your main photo features the closed pool. I suggest swapping it for the 'Fireplace Lobby' image and updating the description to feature 'Autumn Getaways'.

- Promotion (Market Scout): Since flight capacity from domestic hubs is stable, we should launch a 'Stay 3, Pay 2' campaign for weekenders.

Shall I execute the rate change and photo swap for you now?"

### Plan, work, solution ###

**Phase 1: Broker Intent Parsing (The Plan)**

The Broker receives the user's natural language and converts it into a structured Execution Plan.

**Phase 2: Agent Execution (The Work)**

The Broker sends requests to the agents. 
    - Market Scount Response
    - Revenue Strategist Response
    - Content Response

**Phase 3: Broker Synthesis (The Solution)**

The Broker aggregates the above responses into a final strategic plan. This is used to generate the natural language response for the user and prepare the API calls if the user says "Yes".

### How does this deliver value ###

By structuring the agentic network this way, the network is:

**Auditable:** You can see exactly why the broker recommended a price drop (because of the event_change_detected logic).

**Actionable:** The JSON in Phase 3 is ready to be sent to the Ofishal API (PUT /rates, PUT /content) the moment the user approves.

**Modular:** You can swap out the "Revenue Agent" for a different model or provider without breaking the Broker's logic.

### Verify LangGraph successfully Installed

````python
import importlib.metadata
import subprocess
import sys


def ensure_packages(package_list):
    """
    Iterates through a list of packages, checking if they are installed.
    If a package is missing, it attempts to install it via pip.
    """
    for package_name in package_list:
        print(f"--- Checking {package_name} ---")
        try:
            version = importlib.metadata.version(package_name)
            print(f"{package_name} is already installed. Version: {version}\n")
        except importlib.metadata.PackageNotFoundError:
            print(f"{package_name} is NOT installed. Attempting to install now...")
            try:
                # Use subprocess to run pip install
                subprocess.check_call([sys.executable, "-m", "pip", "install", package_name])
                print(f"Successfully installed {package_name}.")
                
                # After installation, attempt to get the version again
                version = importlib.metadata.version(package_name)
                print(f"{package_name} version: {version}\n")
            except subprocess.CalledProcessError as e:
                print(f"Failed to install {package_name}. Error: {e}\n")
            except Exception as e:
                print(f"An unexpected error occurred during installation: {e}\n")

packages_to_install = ['langgraph','langchain','langchain-openai']
ensure_packages(packages_to_install)
````

### Agents and Broker Mockup

````python
import operator
import logging
import sys
import json
import xml.etree.ElementTree as ET
from typing import Annotated, List, TypedDict, Dict
import threading
import queue
import time
import re
import tkinter as tk
from tkinter import ttk, scrolledtext

import langchain
from langgraph.graph import StateGraph, END
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# --- CONFIGURATION ---
config_xml_file = "config_agency.xml"
# config_xml_file = "config_travel.xml"
langchain.debug = False
logging.basicConfig(stream=sys.stdout, level=logging.WARNING, force=False)
logging.getLogger().setLevel(logging.WARNING)

# --- API KEY (never hard-coded) --------------------------------------------
# Provide the key any of three ways; this cell checks all three:
#   1. .env file next to this notebook:  OPENAI_API_KEY=sk-...   (git-ignored)
#   2. Shell environment:                export OPENAI_API_KEY=sk-...
#   3. Colab:  key icon in the left sidebar, secret named OPENAI_API_KEY
import os

try:                                      # 1. .env file
    from dotenv import load_dotenv
    load_dotenv()
except ModuleNotFoundError:
    pass

try:                                      # 3. Colab secrets
    from google.colab import userdata
    if not os.environ.get("OPENAI_API_KEY"):
        os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
except Exception:
    pass

MY_OPENAI_KEY = os.environ.get("OPENAI_API_KEY")
if not MY_OPENAI_KEY:
    raise RuntimeError(
        "OPENAI_API_KEY is not set. Create a .env file next to this notebook "
        "containing OPENAI_API_KEY=sk-..., or export it in your shell."
    )
llm = ChatOpenAI(model="gpt-4o", temperature=0, openai_api_key=MY_OPENAI_KEY)

# --- XML CONFIGURATION LOADER ---
def load_config(xml_path: str) -> dict:
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    config = {
        "prompts": {},
        "scenarios": [],
        "default_scenario": 1,
        "agents": {}
    }

    config["prompts"]["router"] = root.find('./prompts/router_prompt').text.strip()
    config["prompts"]["synthesis"] = root.find('./prompts/synthesis_prompt').text.strip()

    for sc in root.findall('./scenarios/scenario'):
        sc_id = int(sc.get('id'))
        if sc.get('default') == 'true':
            config["default_scenario"] = sc_id
        keywords = [k.text.strip().lower() for k in sc.findall('keyword')]
        config["scenarios"].append((sc_id, keywords))

    for ag in root.findall('./agents/agent'):
        ag_id = ag.get('id')
        mock_data_node = ag.find('mock_data')
        
        agent_data = {
            "name": ag.get('name'),
            "description": ag.find('description').text.strip(),
            "default_mock": mock_data_node.get('default', ''),
            "mocks": {}
        }
        
        for resp in mock_data_node.findall('response'):
            agent_data["mocks"][int(resp.get('scenario_id'))] = resp.text.strip()
            
        config["agents"][ag_id] = agent_data

    return config

# Assumes you saved the consulting firm XML as config.xml
app_config = load_config(config_xml_file)

# --- GUI COMMUNICATION UTILITIES ---
gui_queue = queue.Queue()
human_response_event = threading.Event()
human_response_value = ""
GUI_MODE = False

def log_update(message, msg_type="log"):
    """Sends messages to the GUI or prints to CLI if GUI is off."""
    if GUI_MODE:
        gui_queue.put({"type": msg_type, "msg": message})
    else:
        if msg_type in ["chat", "status"]:
            print(f"\n[{msg_type.upper()}] {message}")

# --- 1. Define State ---
def merge_dicts(left: Dict, right: Dict) -> Dict:
    if not left: left = {}
    if not right: right = {}
    return {**left, **right}

class AgentState(TypedDict):
    input: str
    scenario_id: int
    broker_plan: List[str]
    agent_outputs: Annotated[Dict[str, str], merge_dicts]
    final_response: str
    requires_approval: bool
    approval_decision: str

# --- 2. Dynamic Agent Node Generator ---
def create_agent_node(agent_id: str, agent_config: dict):
    def agent_node(state: AgentState):
        scenario = state['scenario_id']
        mocked_response = agent_config["mocks"].get(scenario, agent_config["default_mock"])
        
        log_update(f"Activating specialized agent: {agent_config['name']}...", "status")
        log_update(f"Agent '{agent_config['name']}' returned data: {mocked_response}", "log")
        
        return {"agent_outputs": {agent_config["name"]: mocked_response}}
    return agent_node

# --- 3. Define Core Nodes ---
def broker_node(state: AgentState):
    user_input = state['input']
    outputs = state.get('agent_outputs', {})
    decision = state.get('approval_decision')

    if decision == "REJECTED":
        msg = "❌ **ACTION CANCELLED:** The human operator rejected the changes."
        state['final_response'] = msg
        state['requires_approval'] = False
        log_update(msg, "chat")
        log_update("Broker: Process aborted by user.", "log")
        return state

    if outputs:
        log_update("Broker: Synthesizing agent data and determining risk...", "status")
        raw_prompt = app_config["prompts"]["synthesis"]
        prompt = raw_prompt.replace("{user_input}", user_input).replace("{agent_data}", json.dumps(outputs))
        
        response = llm.invoke([SystemMessage(content=prompt)])
        content = response.content
        
        state['final_response'] = content
        state['requires_approval'] = "RISK: YES" in content.upper()
        
        log_update("Broker: Synthesis complete. Evaluated risk.", "log")
        
        if not state['requires_approval']:
            log_update(content, "chat")
            log_update("Ready", "status")
        return state

    log_update("Broker: Analyzing intent...", "status")
    agent_list_str = "\n".join([f"- {a_id} ({a_data['description']})" for a_id, a_data in app_config["agents"].items()])
    
    raw_prompt = app_config["prompts"]["router"]
    system_prompt = raw_prompt.replace("{agent_list}", agent_list_str)
    
    routing_response = llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content=user_input)
    ])
    
    cleaned_plan = routing_response.content.replace("```json", "").replace("```", "").strip()
    try:
        plan = json.loads(cleaned_plan)
        plan = [agent for agent in plan if agent in app_config["agents"]]
    except:
        plan = [list(app_config["agents"].keys())[0]] 
        
    state['broker_plan'] = plan
    log_update(f"Broker: Delegating tasks to {', '.join(plan)}", "status")
    log_update(f"Delegated to: {plan}", "log")
    return state

def human_approval_node(state: AgentState):
    global human_response_value
    proposal = state['final_response']
    
    log_update("⚠️ Human Approval Required. Reviewing proposal...", "status")
    log_update(f"PROPOSAL REVIEW:\n{proposal}", "log")
    
    if GUI_MODE:
        log_update(f"🛑 **APPROVAL REQUIRED** 🛑\n\n{proposal}\n\n**Do you authorize this change? (yes/no)**", "chat")
        log_update("requires_input", "system")
        
        human_response_event.clear()
        human_response_event.wait() 
        user_choice = human_response_value.lower()
    else:
        print("\n🛑 STOP! HUMAN APPROVAL REQUIRED 🛑\n" + "-" * 40)
        print(f"PROPOSAL:\n{proposal}\n" + "-" * 40)
        user_choice = input(">> Do you authorize this change? (yes/no): ").lower()
    
    if "yes" in user_choice or "y" in user_choice:
        log_update("Human Approved. Executing...", "log")
        msg = proposal + "\n\n✅ **(STATUS: EXECUTED BY USER)**"
        if GUI_MODE: 
            log_update(msg, "chat")
            log_update("Ready", "status")
        return {"approval_decision": "APPROVED", "final_response": msg}
    else:
        log_update("Human Rejected.", "log")
        return {"approval_decision": "REJECTED"}

# --- 4. Build Graph Dynamically ---
workflow = StateGraph(AgentState)

workflow.add_node("broker", broker_node)
workflow.add_node("human_approval", human_approval_node)

for agent_id, agent_data in app_config["agents"].items():
    workflow.add_node(agent_id, create_agent_node(agent_id, agent_data))

def router(state: AgentState):
    if state.get('approval_decision'):
        return END
    if state.get('requires_approval'):
        return "human_approval"
    if state.get('final_response'):
        return END
    return state.get('broker_plan', [])

workflow.set_entry_point("broker")

possible_edges = list(app_config["agents"].keys()) + ["human_approval", END]
workflow.add_conditional_edges("broker", router, possible_edges)

for agent_id in app_config["agents"].keys():
    workflow.add_edge(agent_id, "broker")
    
workflow.add_edge("human_approval", "broker") 

app = workflow.compile()

# --- 5. Runner Functions ---
def determine_scenario_id(prompt: str) -> int:
    p = prompt.lower()
    for sc_id, keywords in app_config["scenarios"]:
        if any(kw in p for kw in keywords):
            return sc_id
    return app_config["default_scenario"]

def run_scenario(user_prompt: str):
    """Headless CLI runner for testing without the GUI"""
    print(f"\n--- NEW USER REQUEST ---\nUSER: {user_prompt}")
    inferred_id = determine_scenario_id(user_prompt)
    inputs = {"input": user_prompt, "scenario_id": inferred_id, "agent_outputs": {}}
    result = app.invoke(inputs)
    print(f"\nAI RESPONSE (Scenario {inferred_id} Data):\n{result['final_response']}")


# --- 6. Desktop UI Application ---

# Easily configurable UI constants
UI_THEME = {
    "font_family": "Helvetica",
    "font_size_chat": 13,
    "font_size_log": 11,
    "bg_color": "#F4F6F9",        # Light grayish-blue background
    "chat_bg": "#FFFFFF",         # White chat background
    "text_color": "#2C3E50",      # Dark slate text
    "ai_label_color": "#E74C3C",  # Distinctive Red/Orange for AI
    "you_label_color": "#2980B9", # Distinctive Blue for User
    "log_bg": "#1E1E1E",          # Dark background for traces
    "log_fg": "#4CAF50"           # Terminal green for trace text
}

class AgentClientApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Agentic AI")
        self.root.geometry("900x700")
        
        # Configure overall style
        style = ttk.Style()
        style.theme_use('clam')
        style.configure("TNotebook", background=UI_THEME["bg_color"])
        style.configure("TFrame", background=UI_THEME["bg_color"])
        
        self.awaiting_approval = False
        
        self.notebook = ttk.Notebook(root)
        self.notebook.pack(expand=True, fill='both')
        
        self.chat_frame = ttk.Frame(self.notebook)
        self.log_frame = ttk.Frame(self.notebook)
        
        self.notebook.add(self.chat_frame, text='   How can I help?   ')
        self.notebook.add(self.log_frame, text= '   Agent Trace Log   ')
        
        self.setup_chat_tab()
        self.setup_log_tab()
        
        self.poll_queue()

    def setup_chat_tab(self):
        # Status Bar
        self.status_var = tk.StringVar()
        self.status_var.set("Ready")
        status_bar = tk.Label(
            self.chat_frame, 
            textvariable=self.status_var, 
            fg=UI_THEME["you_label_color"], 
            bg=UI_THEME["bg_color"],
            anchor="w", 
            font=(UI_THEME["font_family"], 10, "bold")
        )
        status_bar.pack(side=tk.TOP, fill=tk.X, padx=10, pady=5)
        
        # Chat History Window
        self.chat_history = scrolledtext.ScrolledText(
            self.chat_frame, 
            wrap=tk.WORD, 
            state='disabled', 
            font=(UI_THEME["font_family"], UI_THEME["font_size_chat"]),
            bg=UI_THEME["chat_bg"],
            fg=UI_THEME["text_color"],
            padx=15,
            pady=15
        )
        self.chat_history.pack(expand=True, fill='both', padx=10, pady=(0, 10))
        
        # Configure Font Tags for Markdown and Highlights
        self.chat_history.tag_config("normal", font=(UI_THEME["font_family"], UI_THEME["font_size_chat"]))
        self.chat_history.tag_config("bold", font=(UI_THEME["font_family"], UI_THEME["font_size_chat"], "bold"))
        self.chat_history.tag_config("ai_label", foreground=UI_THEME["ai_label_color"], font=(UI_THEME["font_family"], UI_THEME["font_size_chat"], "bold"))
        self.chat_history.tag_config("you_label", foreground=UI_THEME["you_label_color"], font=(UI_THEME["font_family"], UI_THEME["font_size_chat"], "bold"))
        
        # Input Area
        input_frame = tk.Frame(self.chat_frame, bg=UI_THEME["bg_color"])
        input_frame.pack(fill=tk.X, padx=10, pady=(0, 10))
        
        self.user_input = tk.Entry(
            input_frame, 
            font=(UI_THEME["font_family"], UI_THEME["font_size_chat"]),
            relief=tk.FLAT,
            highlightbackground="#CCCCCC",
            highlightthickness=1
        )
        self.user_input.pack(side=tk.LEFT, expand=True, fill=tk.X, padx=(0, 10), ipady=5)
        self.user_input.bind("<Return>", lambda event: self.send_message())
        
        self.send_btn = tk.Button(
            input_frame, 
            text="Send", 
            command=self.send_message, 
            width=12, 
            bg="#27AE60", 
            fg="black", 
            font=(UI_THEME["font_family"], 11, "bold"),
            relief=tk.FLAT,
            cursor="hand2"
        )
        self.send_btn.pack(side=tk.RIGHT, ipady=3)

    def setup_log_tab(self):
        self.log_history = scrolledtext.ScrolledText(
            self.log_frame, 
            wrap=tk.WORD, 
            state='disabled', 
            font=("Courier", UI_THEME["font_size_log"]), 
            bg=UI_THEME["log_bg"], 
            fg=UI_THEME["log_fg"],
            padx=10,
            pady=10
        )
        self.log_history.pack(expand=True, fill='both', padx=10, pady=10)

    def append_log_text(self, text, color=None):
        self.log_history.config(state='normal')
        if color:
            self.log_history.insert(tk.END, text + "\n", color)
            self.log_history.tag_config(color, foreground=color)
        else:
            self.log_history.insert(tk.END, text + "\n")
        self.log_history.yview(tk.END)
        self.log_history.config(state='disabled')

    def insert_markdown_text(self, widget, text, default_tag="normal"):
        """Parses **bold** markdown and inserts into Tkinter text widget."""
        widget.config(state='normal')
        # Split by **...** matching across lines if necessary
        parts = re.split(r'(\*\*.*?\*\*)', text, flags=re.DOTALL)
        
        for part in parts:
            if part.startswith('**') and part.endswith('**'):
                clean_text = part[2:-2] # Strip the asterisks
                widget.insert(tk.END, clean_text, "bold")
            else:
                widget.insert(tk.END, part, default_tag)
                
        widget.insert(tk.END, "\n")
        widget.yview(tk.END)
        widget.config(state='disabled')

    def poll_queue(self):
        try:
            while True:
                msg_data = gui_queue.get_nowait()
                msg_type = msg_data["type"]
                msg = msg_data["msg"]
                
                if msg_type == "status":
                    self.status_var.set(msg)
                    self.append_log_text(f"[STATUS] {msg}", "#F1C40F") # Yellowish
                elif msg_type == "log":
                    self.append_log_text(f"[TRACE] {msg}", UI_THEME["log_fg"])
                elif msg_type == "chat":
                    self.chat_history.config(state='normal')
                    self.chat_history.insert(tk.END, "\nAI:\n", "ai_label")
                    self.insert_markdown_text(self.chat_history, msg, "normal")
                elif msg_type == "system" and msg == "requires_input":
                    self.awaiting_approval = True
                    self.send_btn.config(text="Approve (Y/N)", bg="#F39C12", fg="black") # Orange
        except queue.Empty:
            pass
        finally:
            self.root.after(100, self.poll_queue) 

    def send_message(self):
        user_text = self.user_input.get().strip()
        if not user_text: return
        
        self.user_input.delete(0, tk.END)
        
        # Insert User Text
        self.chat_history.config(state='normal')
        self.chat_history.insert(tk.END, "\nYou: ", "you_label")
        self.chat_history.insert(tk.END, user_text + "\n", "normal")
        self.chat_history.yview(tk.END)
        self.chat_history.config(state='disabled')
        
        if self.awaiting_approval:
            global human_response_value
            human_response_value = user_text
            self.awaiting_approval = False
            self.send_btn.config(text="Send", bg="#27AE60", fg="black")
            human_response_event.set()
            self.status_var.set("Processing human feedback...")
        else:
            self.status_var.set("Analyzing intent and routing to agents...")
            self.append_log_text(f"\n--- NEW REQUEST: {user_text} ---", "#FFFFFF")
            threading.Thread(target=self.run_graph_thread, args=(user_text,), daemon=True).start()

    def run_graph_thread(self, user_prompt):
        inferred_id = determine_scenario_id(user_prompt)
        inputs = {"input": user_prompt, "scenario_id": inferred_id, "agent_outputs": {}}
        app.invoke(inputs)

def launch_app():
    global GUI_MODE
    GUI_MODE = True
    root = tk.Tk()
    
    root.grid_columnconfigure(0, weight=1)
    root.grid_rowconfigure(0, weight=1)
    
    client = AgentClientApp(root)
    root.mainloop()
    GUI_MODE = False

# --- EXECUTION ---
if __name__ == "__main__":
    # To launch the desktop app, uncomment the next line:
    launch_app()
    
    # To run headlessly in the terminal/notebook instead, use this:
    # run_scenario("There is massive construction happening outside.")
````
